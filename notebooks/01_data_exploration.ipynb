```python
resolution_by_priority = (
    df_clean[df_clean["is_resolved"]]
    .groupby("issue_priority")
    ["resolution_time_hours"]
    .agg(
        ticket_count="count",
        median_hours="median",
        average_hours="mean"
    )
    .reset_index()
)

resolution_by_priority.sort_values("median_hours")
```

```python
workflow_resolution = (
    df_clean[df_clean["is_resolved"]]
    .groupby("workflow_complexity")["resolution_time_hours"]
    .agg(
        ticket_count="count",
        median_hours="median",
        mean_hours="mean"
    )
    .reset_index()
)

workflow_resolution.sort_values("median_hours")
```
